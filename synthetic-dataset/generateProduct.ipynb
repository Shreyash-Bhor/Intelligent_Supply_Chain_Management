```python
for i in range(800) :

    category = random.choice(categories)

    name = random.choice(product_catalog[category])

    product = {
        "name" : name,
        "sku" : name.upper()[:3],
        "id" : str(uuid.uuid4()),
        "isActive" : random.choices([True, False], weights = [75, 25], k = 1)[0],
        "category" : category,
        "imageUrl" : category_url[category]
    }

    data.append(product)
```

```python
df = pd.DataFrame(data)
```
