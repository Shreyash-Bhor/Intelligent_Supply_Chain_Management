```python
import pandas as pd
import random
from faker import Faker
import uuid
```

```python
fake = Faker()
```

```python
data = []
```

```python
for i in range(20000) : 
     
    name = fake.name()

    username = name.lower().replace(" ", "")

    email = f"{username}@gmail.com"

    customer = {
        "isActive" : True,
        "isEmailVerified" : True,
        "name" : name,
        "password" : "$2b$12$eHMTFvjqxcvWkK5eR8IrmutMXADcz/3mdAEYTVgFWHX/.bvEgW8F2",
        "email" : email,
        "id" : str(uuid.uuid4()),
    }

    data.append(customer)
```

```python
df = pd.DataFrame(data)
```
