```python
import pandas as pd
```

```python
# Load dataset
df = pd.read_csv(r"c:\Users\DELL\Desktop\Product.csv")
```

```python
# Display number of rows and columns
df.shape
```

```python
# Create data profiling report
data_profiling = pd.DataFrame ({
    "columns" : df.columns,
    "dtype" : df.dtypes.values,
    "unique_values" : df.nunique().values
})

data_profiling
```

```python
# Create a report for missing values
missing_values = df.isnull().sum()

missing_percentage = (missing_values / len(df)) * 100
 
missing_report = pd.DataFrame ({
    "columns" : df.columns,
    "missing_values" : missing_values.values,
    "missing_percentage" : missing_percentage.values
})

missing_report
```

```python
# Check for duplicate rows in the dataset
df.duplicated().sum()
```

```python
# Check for duplicate values in the "sku" column
df["sku"].duplicated().sum()
```

```python
# Generate unique skus by appending numbers to duplicate skus
count = df.groupby("sku").cumcount()

df["sku"] = df["sku"] + count.replace(0, "").astype(str)
```

```python
# Verify that all skus are now unique
df["sku"].duplicated().sum()
```

```python
# Save the cleaned dataset to a new CSV file
df.to_csv(r"c:\Users\DELL\Desktop\updated_product.csv", index = False)
```
