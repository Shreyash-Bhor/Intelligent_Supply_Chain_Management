```python
import pandas as pd
import random 
import uuid
from faker import Faker
```

```python
products = pd.read_csv(r"C:\Users\DELL\Desktop\product.csv", encoding = "latin1")
```

```python
category_price_range = {
    "Electronics" : (2000, 250000),
    "Grocery" : (20, 2500),
    "Beverages" : (20, 500),
    "Fashion" : (300, 15000),
    "Fitness" : (50, 5000),
    "Stationery" : (10, 2000),
    "Sports" : (500, 30000),
    "Toys" : (200, 10000)
}
```

```python
product_price = []
```

```python
endings = [49, 79, 89, 99]

for category, group in products.groupby("category"):

    # Shuffle products
    group = group.sample(frac=1).reset_index(drop=True)

    total_products = len(group)

    # 20-60-20 Distribution
    budget_count = int(total_products * 0.20)
    mid_count = int(total_products * 0.60)
    premium_count = total_products - budget_count - mid_count

    # Category Price Range
    min_price, max_price = category_price_range[category]

    # Divide the price range into three equal segments
    budget_min = min_price
    budget_max = min_price + (max_price - min_price) // 3

    mid_min = budget_max + 1
    mid_max = min_price + 2 * (max_price - min_price) // 3

    premium_min = mid_max + 1
    premium_max = max_price

    # Generate prices
    for index, row in group.iterrows():

        if index < budget_count:
            segment_min = budget_min
            segment_max = budget_max

        elif index < budget_count + mid_count:
            segment_min = mid_min
            segment_max = mid_max

        else:
            segment_min = premium_min
            segment_max = premium_max

        # Generate random price
        price = random.randint(segment_min, segment_max)

        # Remove last two digits
        price = price - (price % 100)

        # Add realistic retail ending
        price += random.choice(endings)

        # Ensure price stays inside the segment range
        if price > segment_max:
            price -= 100

        if price < segment_min:
            price += 100

        product_price.append({
            "id": str(uuid.uuid4()),
            "productId": row["id"],
            "price": price,
            "currency": "INR"
        })
```

```python
price_df = pd.DataFrame(product_price)
```

```python
price_df.to_csv("product_price.csv", index = False)
```
