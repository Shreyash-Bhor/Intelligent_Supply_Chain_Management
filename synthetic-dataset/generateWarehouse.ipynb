```python
import pandas as pd
import random
from faker import Faker
import uuid
```

```python
fake = Faker()
```

```python
data = []
```

```python
city = ["Mumbai", "Pune", "Ahmedabad", "Delhi", "Jaipur", "Lucknow", "Kolkata", "Bhubaneswar", "Hyderabad", "Bengaluru", "Chennai", "Kochi", "Nagpur", "Indore", "Guwahati", "Chandigarh", "Patna", "Ranchi", "Visakhapatnam", "Coimbatore"]

selected_cities = random.sample(city, k = 20)
```

```python
for i in range(20) : 

    warehouse =  {
        "id" : str(uuid.uuid4()),
        "name" : f"Warehouse {i + 1}",
        "city" : selected_cities[i],
        "pincode" : str(422001 + i),
        "isActive" : True
    }

    data.append(warehouse)
```

```python
df = pd.DataFrame(data)
```

```python
df.to_csv("warehouse.csv", index = False)
```
